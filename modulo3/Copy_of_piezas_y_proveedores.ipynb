Creating Tables for: Providers / Provides / Pieces

```python
%%sql

CREATE TABLE IF NOT EXISTS Pieces(
    Code INTEGER PRIMARY KEY,
    Name TEXT
);

CREATE TABLE IF NOT EXISTS Providers(
   Code TEXT PRIMARY KEY,
    Name TEXT
);

CREATE TABLE IF NOT EXISTS Provides(
   Piece INTEGER,
   Provider TEXT,
   Price INTEGER,
   PRIMARY KEY (Piece, Provider),
   FOREIGN KEY (Piece) REFERENCES Pieces(Code),
   FOREIGN KEY (Provider) REFERENCES Providers(Code)
);
```

5.3 Obtener el precio promedio de cada pieza (mostrar únicamente el código de la pieza y el precio promedio).

```python
%%sql

SELECT
    Pieces.Code,
    AVG(Provides.Price) AS AvgPrice
FROM Pieces
JOIN Provides
    ON Pieces.Code = Provides.Piece
GROUP BY Pieces.Code;
```

5.4 Obtener los nombres de todos los proveedores que suministran la pieza 1.

```python
%%sql

SELECT Providers.Name
FROM Providers
JOIN Provides
    ON Providers.Code = Provides.Provider
WHERE Provides.Piece = 1;
```

5.5 Seleccionar el nombre de las piezas proporcionadas por el proveedor con código "HAL".

```python
%%sql

SELECT Pieces.Name
FROM Pieces
JOIN Provides
    ON Pieces.Code = Provides.Piece
WHERE Provides.Provider = 'HAL';
```

5.6
-- Para cada pieza, encontrar la oferta más cara de esa pieza e incluir el nombre de la pieza,
-- el nombre del proveedor y el precio
-- (tenga en cuenta que puede haber dos proveedores que suministren la misma pieza al precio más caro).

```python
%%sql

SELECT
    Pieces.Name AS Piece,
    Providers.Name AS Provider,
    Provides.Price
FROM Provides
JOIN Pieces
    ON Pieces.Code = Provides.Piece
JOIN Providers
    ON Providers.Code = Provides.Provider
WHERE Provides.Price = (
    SELECT MAX(Price)
    FROM Provides AS P2
    WHERE P2.Piece = Provides.Piece
);
```
