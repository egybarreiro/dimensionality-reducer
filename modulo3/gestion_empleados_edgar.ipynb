Creating Departments Table & Employees Table

```python
%%sql
DROP TABLE IF EXISTS Employees;
DROP TABLE IF EXISTS Departments;

CREATE TABLE Departments(
    code INTEGER PRIMARY KEY,
    name TEXT NOT NULL,
    budget INTEGER NOT NULL
);

CREATE TABLE Employees(
    ssn TEXT PRIMARY KEY,
    fname TEXT NOT NULL,
    lname TEXT NOT NULL,
    dept INTEGER,
    FOREIGN KEY(dept) REFERENCES Departments(code)
);
```

2.9 Selecciona la cantidad de empleados en cada departamento (solo mostrar código del departamento y número de empleados).

2.10 Selecciona todos los datos de empleados, incluyendo los datos de su departamento.

```python
%%sql
SELECT *
FROM Employees
JOIN Departments
ON Employees.dept = Departments.code
```

2.11 Selecciona el nombre y apellido de cada empleado junto con el nombre y presupuesto de su departamento.

```python
%%sql
SELECT fname, lname, name, budget
FROM Employees
JOIN Departments
ON Employees.dept = Departments.code
```

2.12 Selecciona el nombre y apellido de empleados que trabajan en departamentos con presupuesto mayor a $60,000.

```python
%%sql
SELECT fname, lname
FROM Employees
JOIN Departments
ON Employees.dept = Departments.code
WHERE Departments.budget > 60000
```

2.13 Selecciona los departamentos con presupuesto mayor al promedio de todos los departamentos.

```python
%%sql
SELECT name, budget
FROM Departments
WHERE budget > (SELECT AVG(budget) FROM Departments)
```

2.14 Selecciona los nombres de departamentos con más de dos empleados.

```python
%%sql
SELECT Departments.name
FROM Employees
JOIN Departments
ON Employees.dept = Departments.code
GROUP BY Departments.name
HAVING COUNT(*) > 2
```

2.15 Muy importante - Selecciona el nombre y apellido de empleados que trabajan en el departamento con el segundo presupuesto más bajo.

```python
%%sql
SELECT fname, lname
FROM Employees
JOIN Departments ON Employees.dept = Departments.code
WHERE Departments.code = (
    SELECT code
    FROM Departments
    ORDER BY budget ASC
    LIMIT 1 OFFSET 1
);
```
